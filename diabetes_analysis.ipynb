## Phase 1: Data Ingestion & Clinical Sanitation

This notebook performs data ingestion, cleaning, and preprocessing on the
"Diabetes 130-US Hospitals (1999–2008)" dataset.  
The objective of Phase 1 is to transform raw clinical data into a reliable
analytical dataset suitable for readmission analysis.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

We import pandas for data manipulation, numpy for numerical operations, and visualization libraries for later phases.

```python
# Load the diabetic data
df = pd.read_csv('diabetic_data.csv')

# Display first few rows
print("First 5 rows of the dataset:")
df.head()
```

We load the clinical dataset into a pandas DataFrame to begin our analysis.

```python
# Check dataset dimensions
print(f"Dataset shape: {df.shape}")
print(f"Total rows: {df.shape[0]}, Total columns: {df.shape[1]}")

# View all column names
print("\nColumn names:")
print(df.columns.tolist())

# Get data types and non-null counts
print("\nData types and non-null information:")
df.info()

# Get summary statistics for numerical columns
print("\nSummary statistics:")
df.describe()
```

We perform an initial audit to understand the structure, size, and data types of our dataset. This helps identify potential issues before cleaning

```python
# Check for '?' in the dataset
print("Checking for '?' values in each column:")
for col in df.columns:
    question_mark_count = (df[col] == '?').sum()
    if question_mark_count > 0:
        print(f"{col}: {question_mark_count} occurrences")

# Replace '?' with NaN
df.replace('?', np.nan, inplace=True)

# Verify replacement
print("\nMissing values after replacement:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

The dataset uses '?' to represent missing values. We replace these with standard NaN values for proper handling in pandas.

```python
# Calculate missing percentage for weight column
weight_missing_pct = (df['weight'].isnull().sum() / len(df)) * 100
print(f"Weight column missing percentage: {weight_missing_pct:.2f}%")

# Decision logic
if weight_missing_pct > 90:
    print(f"\nDropping 'weight' column due to {weight_missing_pct:.2f}% missingness")
    df.drop('weight', axis=1, inplace=True)
    print("Weight column dropped successfully")
else:
    print(f"\nWeight column retained (missing: {weight_missing_pct:.2f}%)")

# Verify
print(f"\nDataset shape after weight column handling: {df.shape}")
```

The weight column has excessive missingness (>90%). Following data quality best practices, we drop columns with such high missing rates as they provide minimal analytical value. This is documented as a 'Data Quality Limitation' in our final report.

```python
# Load the IDs mapping file
ids_mapping = pd.read_csv('IDs_mapping.csv')

print("IDs Mapping file structure:")
print(ids_mapping.head(20))

# The file contains three sections separated by empty rows
# We need to separate them for admission_type, discharge_disposition, and admission_source

# Create separate dataframes for each mapping type
# Find the empty rows to split the data
empty_rows = ids_mapping[ids_mapping.isnull().all(axis=1)].index.tolist()
print(f"\nEmpty row indices: {empty_rows}")

# Split into three sections
admission_type_df = ids_mapping.iloc[:empty_rows[0]].copy()
discharge_disp_df = ids_mapping.iloc[empty_rows[0]+1:empty_rows[1]].copy()
admission_source_df = ids_mapping.iloc[empty_rows[1]+1:].copy()

# Clean column names and create dictionaries
admission_type_df.columns = ['admission_type_id', 'admission_type_desc']
discharge_disp_df.columns = ['discharge_disposition_id', 'discharge_disposition_desc']
admission_source_df.columns = ['admission_source_id', 'admission_source_desc']

# Remove any remaining NaN rows
admission_type_df = admission_type_df.dropna()
discharge_disp_df = discharge_disp_df.dropna()
admission_source_df = admission_source_df.dropna()

print("\nAdmission Type Mappings:")
print(admission_type_df)
print("\nDischarge Disposition Mappings:")
print(discharge_disp_df.head(10))
print("\nAdmission Source Mappings:")
print(admission_source_df.head(10))
```

We load the IDs_mapping.csv file which contains descriptions for coded values in our dataset. This will help us translate numerical IDs into meaningful categories.

```python
# Identify "Expired" discharge disposition IDs
expired_ids = discharge_disp_df[
    discharge_disp_df['discharge_disposition_desc'].str.contains('Expired', case=False, na=False)
]['discharge_disposition_id'].values

print(f"Discharge disposition IDs for 'Expired' patients: {expired_ids}")

# Count deceased patients in dataset
deceased_count = df[df['discharge_disposition_id'].isin(expired_ids)].shape[0]
total_count = df.shape[0]
deceased_pct = (deceased_count / total_count) * 100

print(f"\nTotal patients in dataset: {total_count}")
print(f"Deceased patients: {deceased_count} ({deceased_pct:.2f}%)")

# Show some examples of deceased patients
print("\nSample of deceased patient records:")
print(df[df['discharge_disposition_id'].isin(expired_ids)].head())
```

Patients who expired (died) during their hospital stay cannot be readmitted. We identify these patients using discharge disposition codes 11, 19, 20, and 21 from the mapping file.

```python
# Store original shape
original_shape = df.shape

# Filter out deceased patients
df_cleaned = df[~df['discharge_disposition_id'].isin(expired_ids)].copy()

# Verify removal
new_shape = df_cleaned.shape
rows_removed = original_shape[0] - new_shape[0]

print(f"Original dataset: {original_shape[0]} rows")
print(f"After removing deceased patients: {new_shape[0]} rows")
print(f"Rows removed: {rows_removed}")
print(f"Percentage removed: {(rows_removed/original_shape[0])*100:.2f}%")

# Update the main dataframe
df = df_cleaned.copy()
```

We remove deceased patients from our analysis as they cannot be readmitted (our target variable). Including them would introduce noise into our readmission prediction model. This filtering improves the methodological rigor of our analysis.

```python
# Check for exact duplicates
duplicate_count = df.duplicated().sum()
print(f"Number of exact duplicate rows: {duplicate_count}")

if duplicate_count > 0:
    print(f"\nRemoving {duplicate_count} duplicate rows...")
    df_before = df.shape[0]
    df = df.drop_duplicates()
    df_after = df.shape[0]
    print(f"Rows before: {df_before}")
    print(f"Rows after: {df_after}")
    print(f"Duplicates removed: {df_before - df_after}")
else:
    print("\nNo exact duplicates found. Dataset is clean.")

print(f"\nFinal dataset shape after Phase 1: {df.shape}")
```

We check for and remove exact duplicate rows, which represent data entry errors. Note that multiple encounters per patient (same patient_nbr, different encounter_id) are valid and should NOT be removed.

```python
# Identify ID columns that should be categorical
id_columns = ['admission_type_id', 'discharge_disposition_id', 'admission_source_id']

# Convert to categorical
for col in id_columns:
    if col in df.columns:
        df[col] = df[col].astype('category')
        print(f"Converted {col} to category type")

# Also convert other categorical text columns
categorical_text_cols = ['race', 'gender', 'age', 'payer_code', 'medical_specialty',
                         'max_glu_serum', 'A1Cresult', 'insulin', 'change', 
                         'diabetesMed', 'readmitted']

for col in categorical_text_cols:
    if col in df.columns:
        df[col] = df[col].astype('category')

print("\nData types after optimization:")
print(df.dtypes)
```

We optimize data types by converting ID and text columns to categorical type. This reduces memory usage and can improve performance for later operations.
